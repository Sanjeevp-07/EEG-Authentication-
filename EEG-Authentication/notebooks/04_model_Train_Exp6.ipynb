```python
def preprocess_recording(vhdr_path):

    raw = mne.io.read_raw_brainvision(
        vhdr_path,
        preload=True,
        verbose=False
    )

    # ------------------------------------------------
    # Check sampling frequency
    # ------------------------------------------------

    fs = raw.info["sfreq"]

    if abs(fs - EXPECTED_FS) > 1e-6:
        raise ValueError(
            f"Expected {EXPECTED_FS} Hz but found {fs} Hz:\n"
            f"{vhdr_path}"
        )

    # ------------------------------------------------
    # Select channels
    # ------------------------------------------------

    missing = [
        ch for ch in CHANNELS
        if ch not in raw.ch_names
    ]

    if missing:
        raise ValueError(
            f"Missing channels {missing} in {vhdr_path}"
        )

    raw.pick(CHANNELS)

    # ------------------------------------------------
    # Bandpass filter
    # ------------------------------------------------

    raw.filter(
        l_freq=LOWCUT,
        h_freq=HIGHCUT,
        method="fir",
        phase="zero",
        verbose=False
    )

    # ------------------------------------------------
    # Get data
    # ------------------------------------------------

    data = raw.get_data()

    # Convert V → µV
    data_uv = data * 1e6

    # ------------------------------------------------
    # Windowing
    # ------------------------------------------------

    n_samples = data_uv.shape[1]

    windows = []
    rejected = 0

    for start in range(
        0,
        n_samples - WINDOW_SAMPLES + 1,
        STEP_SAMPLES
    ):

        end = start + WINDOW_SAMPLES

        window = data_uv[:, start:end]

        # --------------------------------------------
        # NaN / Inf
        # --------------------------------------------

        if not np.isfinite(window).all():
            rejected += 1
            continue

        # --------------------------------------------
        # Frontal artifact rejection
        # --------------------------------------------

        fp1_ptp = np.ptp(window[0])
        fp2_ptp = np.ptp(window[1])

        if (
            fp1_ptp > ARTIFACT_PTP_UV
            or
            fp2_ptp > ARTIFACT_PTP_UV
        ):
            rejected += 1
            continue

        # --------------------------------------------
        # Flatline rejection
        # --------------------------------------------

        channel_std = np.std(
            window,
            axis=1
        )

        if np.any(
            channel_std < MIN_CHANNEL_STD_UV
        ):
            rejected += 1
            continue

        windows.append(window)

    if len(windows) == 0:
        return np.empty(
            (0, 3, WINDOW_SAMPLES),
            dtype=np.float32
        ), rejected

    windows = np.asarray(
        windows,
        dtype=np.float32
    )

    return windows, rejected
```

```python
def hjorth_parameters(signal):

    activity = np.var(signal)

    diff1 = np.diff(signal)
    diff2 = np.diff(diff1)

    var_diff1 = np.var(diff1)
    var_diff2 = np.var(diff2)

    if activity <= 1e-12:
        mobility = 0.0
    else:
        mobility = np.sqrt(
            var_diff1 / activity
        )

    if var_diff1 <= 1e-12:
        complexity = 0.0
    else:
        mobility_diff = np.sqrt(
            var_diff2 / var_diff1
        )

        complexity = (
            mobility_diff / mobility
            if mobility > 1e-12
            else 0.0
        )

    return activity, mobility, complexity
```

```python
def extract_features(window, fs=500):

    features = []
    names = []

    for ch_idx, ch_name in enumerate(CHANNELS):

        signal = window[ch_idx]

        # ------------------------------------------
        # Time-domain
        # ------------------------------------------

        mean_val = np.mean(signal)
        std_val = np.std(signal)
        rms_val = np.sqrt(
            np.mean(signal ** 2)
        )

        skew_val = skew(signal)
        kurt_val = kurtosis(signal)

        activity, mobility, complexity = (
            hjorth_parameters(signal)
        )

        time_features = {
            "mean": mean_val,
            "std": std_val,
            "rms": rms_val,
            "skew": skew_val,
            "kurtosis": kurt_val,
            "hjorth_activity": activity,
            "hjorth_mobility": mobility,
            "hjorth_complexity": complexity
        }

        for name, value in time_features.items():

            names.append(
                f"{ch_name}_{name}"
            )

            features.append(value)

        # ------------------------------------------
        # Welch PSD
        # ------------------------------------------

        freqs, psd = welch(
            signal,
            fs=fs,
            nperseg=min(
                512,
                len(signal)
            ),
            noverlap=None
        )

        # 1–45 Hz
        valid = (
            (freqs >= 1)
            &
            (freqs <= 45)
        )

        freqs_valid = freqs[valid]
        psd_valid = psd[valid]

        total_power = np.trapezoid(
            psd_valid,
            freqs_valid
        )

        # ------------------------------------------
        # Band powers
        # ------------------------------------------

        band_powers = {}

        for band_name, (low, high) in BANDS.items():

            mask = (
                (freqs >= low)
                &
                (freqs < high)
            )

            if np.any(mask):

                power = np.trapezoid(
                    psd[mask],
                    freqs[mask]
                )

            else:

                power = 0.0

            band_powers[band_name] = power

            features.append(
                power
            )

            names.append(
                f"{ch_name}_{band_name}_power"
            )

        # ------------------------------------------
        # Relative bandpower
        # ------------------------------------------

        for band_name in BANDS:

            rel = (
                band_powers[band_name]
                / (total_power + 1e-12)
            )

            features.append(rel)

            names.append(
                f"{ch_name}_{band_name}_relative"
            )

        # ------------------------------------------
        # Log PSD bins
        # ------------------------------------------

        log_psd = np.log10(
            psd_valid + 1e-12
        )

        # 20 evenly distributed PSD features
        psd_bins = np.array_split(
            log_psd,
            20
        )

        for i, b in enumerate(psd_bins):

            value = np.mean(b)

            features.append(value)

            names.append(
                f"{ch_name}_logPSD_bin_{i}"
            )

    return (
        np.asarray(features, dtype=np.float32),
        names
    )
```

```python
def train_subject_svm(
    target_subject_id,
    X_train,
    y_train
):

    # Binary labels
    # Target subject     -> +1
    # Everyone else      -> -1

    y_binary = np.where(
        y_train == target_subject_id,
        1,
        -1
    )

    positive_count = np.sum(
        y_binary == 1
    )

    negative_count = np.sum(
        y_binary == -1
    )

    if positive_count == 0:
        raise ValueError(
            f"No positive samples for subject "
            f"{target_subject_id}"
        )

    if negative_count == 0:
        raise ValueError(
            f"No negative samples for subject "
            f"{target_subject_id}"
        )

    # Standardization

    scaler = StandardScaler()

    X_scaled = scaler.fit_transform(
        X_train
    )

    # PCA

    pca = PCA(
        n_components=PCA_VARIANCE,
        svd_solver="full",
        random_state=SEED
    )

    X_pca = pca.fit_transform(
        X_scaled
    )

    # --------------------------------------------------------
    # Binary SVM
    # --------------------------------------------------------

    svm = SVC(
        kernel=SVM_KERNEL,
        C=SVM_C,
        gamma=SVM_GAMMA,

        # 1 genuine class vs 59 impostor classes.
        
        class_weight="balanced",

        probability=False,
        random_state=SEED
    )

    svm.fit(
        X_pca,
        y_binary
    )

    return {
        "subject_id": target_subject_id,
        "scaler": scaler,
        "pca": pca,
        "svm": svm,
        "n_positive": positive_count,
        "n_negative": negative_count,
        "n_pca_components": X_pca.shape[1]
    }
```

```python
def create_attempts(
    scores,
    K
):

    n_attempts = (
        len(scores) // K
    )

    if n_attempts == 0:
        return np.array([])

    trimmed = scores[
        :n_attempts * K
    ]

    attempts = trimmed.reshape(
        n_attempts,
        K
    ).mean(axis=1)

    return attempts
```

```python
def collect_authentication_scores(
    models,
    X_test,
    y_test,
    K
):

    genuine_scores = []
    impostor_scores = []

    subject_results = {}

    for target_subject_id in tqdm(
        range(len(subjects)),
        desc=f"Evaluating K={K}"
    ):

        model_bundle = models[
            target_subject_id
        ]

        scaler = model_bundle[
            "scaler"
        ]

        pca = model_bundle[
            "pca"
        ]

        svm = model_bundle[
            "svm"
        ]

        # Transform Session 3

        X_test_scaled = scaler.transform(
            X_test
        )

        X_test_pca = pca.transform(
            X_test_scaled
        )

        # SVM decision scores


        all_scores = svm.decision_function(
            X_test_pca
        )

        # Target subject = GENUINE

        genuine_mask = (
            y_test == target_subject_id
        )

        genuine_window_scores = (
            all_scores[
                genuine_mask
            ]
        )

        genuine_attempt_scores = (
            create_attempts(
                genuine_window_scores,
                K
            )
        )


        impostor_attempt_scores = []

        for impostor_subject_id in range(
            len(subjects)
        ):

            if (
                impostor_subject_id
                == target_subject_id
            ):
                continue

            impostor_mask = (
                y_test
                == impostor_subject_id
            )

            subject_impostor_scores = (
                all_scores[
                    impostor_mask
                ]
            )

            attempts = create_attempts(
                subject_impostor_scores,
                K
            )

            if len(attempts) > 0:

                impostor_attempt_scores.extend(
                    attempts.tolist()
                )

        impostor_attempt_scores = np.asarray(
            impostor_attempt_scores,
            dtype=np.float64
        )


        if len(genuine_attempt_scores) > 0:

            genuine_scores.extend(
                genuine_attempt_scores.tolist()
            )

        if len(impostor_attempt_scores) > 0:

            impostor_scores.extend(
                impostor_attempt_scores.tolist()
            )

        subject_results[
            target_subject_id
        ] = {
            "genuine_scores":
                genuine_attempt_scores,

            "impostor_scores":
                impostor_attempt_scores
        }

    genuine_scores = np.asarray(
        genuine_scores,
        dtype=np.float64
    )

    impostor_scores = np.asarray(
        impostor_scores,
        dtype=np.float64
    )

    return (
        genuine_scores,
        impostor_scores,
        subject_results
    )
```

```python
def calculate_eer(
    genuine_scores,
    impostor_scores
):

    if len(genuine_scores) == 0:
        raise ValueError(
            "No genuine scores available."
        )

    if len(impostor_scores) == 0:
        raise ValueError(
            "No impostor scores available."
        )

    # Genuine  = 1
    # Impostor = 0

    y_true = np.concatenate([
        np.ones(
            len(genuine_scores)
        ),

        np.zeros(
            len(impostor_scores)
        )
    ])

    # Scores

    y_score = np.concatenate([
        genuine_scores,
        impostor_scores
    ])

  
    # ROC
 

    fpr, tpr, thresholds = roc_curve(
        y_true,
        y_score
    )

    # FRR

    fnr = 1.0 - tpr

    # EER


    eer_index = np.nanargmin(
        np.abs(
            fpr - fnr
        )
    )

    far_at_eer = fpr[
        eer_index
    ]

    frr_at_eer = fnr[
        eer_index
    ]

    eer = (
        far_at_eer
        + frr_at_eer
    ) / 2.0

    eer_threshold = thresholds[
        eer_index
    ]

    tar_at_eer = tpr[
        eer_index
    ]

    # ROC AUC

    auc = roc_auc_score(
        y_true,
        y_score
    )

    return {
        "EER": eer,
        "EER_%": eer * 100,

        "threshold":
            eer_threshold,

        "FAR":
            far_at_eer,

        "FAR_%":
            far_at_eer * 100,

        "FRR":
            frr_at_eer,

        "FRR_%":
            frr_at_eer * 100,

        "TAR":
            tar_at_eer,

        "TAR_%":
            tar_at_eer * 100,

        "ROC_AUC":
            auc,

        "fpr":
            fpr,

        "tpr":
            tpr,

        "fnr":
            fnr,

        "thresholds":
            thresholds
    }
```

# Consistency and Robustness Evaluation (added cells)

Everything above this cell is unchanged. The cells below only **add** checks that test whether the S1+S2 -> S3 result is consistent:

1. Data integrity and leakage checks
2. Re-run of the same protocol with reusable helpers (must reproduce the original numbers exactly)
3. Three-fold session rotation (S1+S2->S3, S1+S3->S2, S2+S3->S1)
4. Subject-level bootstrap confidence intervals
5. Reliable per-subject analysis (window level, enough samples per subject)
6. Label-permutation negative control (EER must be about 50%)
7. Open-set evaluation (impostors never seen during training)
8. Final PASS / FAIL summary and CSV export


```python
# ============================================================
# CELL 34
# DATA INTEGRITY AND LEAKAGE CHECKS
# ============================================================

CHECKS = []   # (name, status, detail) -> printed again in the final summary


def add_check(name, passed, detail="", informational=False):
    status = "INFO" if informational else ("PASS" if passed else "FAIL")
    CHECKS.append((name, status, detail))
    print(f"[{status}] {name}" + (f"  ->  {detail}" if detail else ""))


print("=" * 80)
print("DATA INTEGRITY AND LEAKAGE CHECKS")
print("=" * 80)

add_check(
    "Feature matrix and metadata have the same number of rows",
    len(X_features) == len(metadata_df),
    f"{len(X_features)} features vs {len(metadata_df)} metadata rows"
)

add_check(
    "No NaN / Inf values in the feature matrix",
    bool(np.isfinite(X_features).all())
)

n_constant = int((X_features.std(axis=0) == 0).sum())
add_check(
    "No constant (zero-variance) feature columns",
    n_constant == 0,
    f"{n_constant} constant columns"
)

add_check(
    "Enrollment rows and authentication rows are disjoint",
    not bool((enrollment_mask & test_mask).any())
)

n_enroll_rows = int(enrollment_mask.sum())
add_check(
    "StandardScaler was fitted on enrollment rows only (not on Session 3)",
    int(models[0]["scaler"].n_samples_seen_) == n_enroll_rows,
    f"scaler saw {int(models[0]['scaler'].n_samples_seen_)} rows, enrollment has {n_enroll_rows}"
)

row_hash = pd.util.hash_pandas_object(
    pd.DataFrame(X_features.round(4)),
    index=False
)
n_duplicate_rows = int(row_hash.duplicated().sum())
add_check(
    "No duplicated feature rows (would indicate duplicated windows / leakage)",
    n_duplicate_rows == 0,
    f"{n_duplicate_rows} duplicated rows"
)

# Determinism: training the same subject model twice must give identical scores
_m1 = train_subject_svm(0, X_enrollment, y_enrollment)
_m2 = train_subject_svm(0, X_enrollment, y_enrollment)


def _score(m, X):
    return m["svm"].decision_function(
        m["pca"].transform(m["scaler"].transform(X))
    )


add_check(
    "Training is deterministic (same data -> identical decision scores)",
    bool(np.allclose(_score(_m1, X_test), _score(_m2, X_test)))
)

# Window coverage per subject and session
ALL_SESSIONS = sorted(metadata_df["session"].unique())

coverage = (
    metadata_df
    .groupby(["subject", "session"])
    .size()
    .unstack(fill_value=0)
    .reindex(index=subjects, columns=ALL_SESSIONS, fill_value=0)
)

print("\nClean windows per subject, per session:")
print(coverage.describe().loc[["count", "mean", "min", "max"]].round(1))

for sess in ALL_SESSIONS:
    empty = coverage.index[coverage[sess] == 0].tolist()
    print(f"\nSubjects with ZERO clean windows in {sess}: {empty}")

add_check(
    "Every subject has at least one clean window in every session",
    bool((coverage > 0).all().all()),
    "subjects without data cannot be evaluated as genuine users in that session",
    informational=True
)

few = coverage.index[coverage.sum(axis=1) < 30].tolist()
print("\nSubjects with fewer than 30 clean windows in total:", few)
```

```python
# ============================================================
# CELL 35
# REUSABLE PROTOCOL HELPERS
# (they call your existing train_subject_svm / create_attempts)
# ============================================================

N_SUBJECTS = len(subjects)

FOLDS = {
    "S1+S2 -> S3": (["ses-session1", "ses-session2"], "ses-session3"),
    "S1+S3 -> S2": (["ses-session1", "ses-session3"], "ses-session2"),
    "S2+S3 -> S1": (["ses-session2", "ses-session3"], "ses-session1"),
}


def get_split(enroll_sessions, test_session):
    enr = metadata_df["session"].isin(enroll_sessions).values
    te = (metadata_df["session"] == test_session).values
    return (
        X_features[enr],
        metadata_df.loc[enr, "subject_id"].values,
        X_features[te],
        metadata_df.loc[te, "subject_id"].values,
    )


def train_models(X_train, y_train, target_ids, desc="Training SVMs"):
    out = {}
    for s in tqdm(target_ids, desc=desc, leave=False):
        out[s] = train_subject_svm(s, X_train, y_train)
    return out


def score_windows(models_, X_eval):
    # NO fitting here, transform only
    return {
        s: b["svm"].decision_function(
            b["pca"].transform(b["scaler"].transform(X_eval))
        )
        for s, b in models_.items()
    }


def build_attempts(window_scores, y_eval, K, impostor_ids=None):
    """Same logic as collect_authentication_scores, but works on any
    subset of target models and any set of impostor subjects."""

    idx = {s: np.where(y_eval == s)[0] for s in range(N_SUBJECTS)}
    pool = range(N_SUBJECTS) if impostor_ids is None else impostor_ids

    gen_all, imp_all, per_subject = [], [], {}

    for target, scores in window_scores.items():

        g = create_attempts(scores[idx[target]], K)

        imp = []
        for other in pool:
            if other == target:
                continue
            a = create_attempts(scores[idx[other]], K)
            if len(a) > 0:
                imp.append(a)

        imp = np.concatenate(imp) if imp else np.array([])

        per_subject[target] = {"genuine": g, "impostor": imp}

        if len(g) > 0:
            gen_all.append(g)
        if len(imp) > 0:
            imp_all.append(imp)

    if not gen_all or not imp_all:
        raise ValueError("No genuine or no impostor attempts available.")

    return np.concatenate(gen_all), np.concatenate(imp_all), per_subject


def eer_auc(genuine, impostor):
    y = np.r_[np.ones(len(genuine)), np.zeros(len(impostor))]
    s = np.r_[genuine, impostor]
    fpr, tpr, _ = roc_curve(y, s)
    fnr = 1 - tpr
    i = np.nanargmin(np.abs(fpr - fnr))
    return (fpr[i] + fnr[i]) / 2.0, roc_auc_score(y, s)


def run_protocol(enroll_sessions, test_session, k_list=K_LIST, desc="Training"):
    X_tr, y_tr, X_te, y_te = get_split(enroll_sessions, test_session)

    counts = np.bincount(y_tr, minlength=N_SUBJECTS)
    eligible = [s for s in range(N_SUBJECTS) if counts[s] >= 1]

    models_ = train_models(X_tr, y_tr, eligible, desc=desc)
    ws = score_windows(models_, X_te)

    out = {
        "enroll_counts": counts,
        "eligible": eligible,
        "y_te": y_te,
        "window_scores": ws,
        "by_K": {},
    }

    for K in k_list:
        g, i, per = build_attempts(ws, y_te, K)
        e, a = eer_auc(g, i)
        out["by_K"][K] = {
            "EER": e, "AUC": a,
            "n_genuine": len(g), "n_impostor": len(i),
            "genuine": g, "impostor": i, "per_subject": per,
        }

    return out


print("Helpers ready.")
print("Protocols:", list(FOLDS))
```

```python
# ============================================================
# CELL 36
# REPRODUCTION CHECK
# The refactored protocol must give EXACTLY the numbers of CELL 26
# ============================================================

print("=" * 80)
print("REPRODUCTION CHECK  (S1+S2 -> S3)")
print("=" * 80)

fold_runs = {}
fold_runs["S1+S2 -> S3"] = run_protocol(*FOLDS["S1+S2 -> S3"], desc="S1+S2 -> S3")
base = fold_runs["S1+S2 -> S3"]

rows, ok = [], True

for _, r in results_df.iterrows():
    K = int(r["K"])
    new = base["by_K"][K]

    d_eer = abs(new["EER"] * 100 - r["EER_%"])
    d_auc = abs(new["AUC"] - r["ROC_AUC"])
    same_counts = (
        new["n_genuine"] == r["n_genuine"]
        and new["n_impostor"] == r["n_impostor"]
    )

    ok = ok and d_eer < 1e-6 and d_auc < 1e-9 and same_counts

    rows.append({
        "K": K,
        "EER_original_%": r["EER_%"],
        "EER_rerun_%": new["EER"] * 100,
        "abs_diff": d_eer,
        "AUC_original": r["ROC_AUC"],
        "AUC_rerun": new["AUC"],
        "counts_match": same_counts,
    })

display(pd.DataFrame(rows))

add_check(
    "Refactored protocol reproduces the original CELL 26 results exactly",
    ok
)
```

```python
# ============================================================
# CELL 37
# THREE-FOLD SESSION ROTATION
# ============================================================

for name in ["S1+S3 -> S2", "S2+S3 -> S1"]:
    fold_runs[name] = run_protocol(*FOLDS[name], desc=name)

fold_rows = []

for name, run in fold_runs.items():
    for K in K_LIST:
        m = run["by_K"][K]
        fold_rows.append({
            "protocol": name,
            "K": K,
            "attempt_seconds": K * WINDOW_SECONDS,
            "n_models": len(run["eligible"]),
            "n_genuine": m["n_genuine"],
            "n_impostor": m["n_impostor"],
            "EER_%": m["EER"] * 100,
            "ROC_AUC": m["AUC"],
        })

fold_results_df = pd.DataFrame(fold_rows)

print("=" * 80)
print("RESULTS PER FOLD")
print("=" * 80)
display(fold_results_df)

summary_k = (
    fold_results_df
    .groupby("K")
    .agg(
        EER_mean=("EER_%", "mean"),
        EER_std=("EER_%", "std"),
        EER_min=("EER_%", "min"),
        EER_max=("EER_%", "max"),
        AUC_mean=("ROC_AUC", "mean"),
        AUC_std=("ROC_AUC", "std"),
    )
    .round(3)
)

print("\nMean +- std over the 3 folds:")
display(summary_k)

plt.figure(figsize=(8, 5))
for name in FOLDS:
    sub = fold_results_df[fold_results_df["protocol"] == name]
    plt.plot(sub["attempt_seconds"], sub["EER_%"], marker="o", label=name)
plt.errorbar(
    [k * WINDOW_SECONDS for k in summary_k.index],
    summary_k["EER_mean"],
    yerr=summary_k["EER_std"],
    color="black", linewidth=2, capsize=4, label="mean +- std"
)
plt.xlabel("Authentication Duration (seconds)")
plt.ylabel("EER (%)")
plt.title("EER vs Authentication Duration, all three folds")
plt.grid(alpha=0.3)
plt.legend()
plt.show()

# ---- automatic checks -------------------------------------------------
mean_eer = summary_k["EER_mean"].values
add_check(
    "Mean EER decreases monotonically as K (authentication time) increases",
    bool(np.all(np.diff(mean_eer) < 0)),
    " -> ".join(f"{v:.2f}%" for v in mean_eer)
)

STABILITY_PP = 3.0
max_std = float(summary_k["EER_std"].max())
add_check(
    f"EER is stable across folds (std <= {STABILITY_PP} percentage points)",
    max_std <= STABILITY_PP,
    f"largest std over K = {max_std:.2f} pp"
)
```

```python
# ============================================================
# CELL 38
# SUBJECT-LEVEL BOOTSTRAP CONFIDENCE INTERVALS
# (resample target subjects with replacement, S1+S2 -> S3)
# ============================================================

N_BOOT = 200

rng = np.random.default_rng(SEED)

boot_rows = []

for K in K_LIST:

    per = base["by_K"][K]["per_subject"]

    targets = [s for s in per if len(per[s]["genuine"]) > 0]

    boot_eers = []

    for _ in tqdm(range(N_BOOT), desc=f"Bootstrap K={K}", leave=False):

        sample = rng.choice(targets, size=len(targets), replace=True)

        g = np.concatenate([per[s]["genuine"] for s in sample])
        i = np.concatenate([per[s]["impostor"] for s in sample])

        boot_eers.append(eer_auc(g, i)[0] * 100)

    boot_eers = np.asarray(boot_eers)

    boot_rows.append({
        "K": K,
        "EER_%": base["by_K"][K]["EER"] * 100,
        "CI95_low_%": np.percentile(boot_eers, 2.5),
        "CI95_high_%": np.percentile(boot_eers, 97.5),
        "boot_std_%": boot_eers.std(ddof=1),
        "n_subjects_with_genuine": len(targets),
    })

bootstrap_df = pd.DataFrame(boot_rows)

print("=" * 80)
print("EER WITH 95% BOOTSTRAP CONFIDENCE INTERVAL (S1+S2 -> S3)")
print("=" * 80)
display(bootstrap_df.round(2))

# CIs should be reasonably tight and ordered
add_check(
    "Bootstrap CI contains the point estimate for every K",
    bool(((bootstrap_df["CI95_low_%"] <= bootstrap_df["EER_%"] + 1e-9)
          & (bootstrap_df["EER_%"] <= bootstrap_df["CI95_high_%"] + 1e-9)).all()),
    "; ".join(
        f"K={int(r.K)}: {r['EER_%']:.2f} [{r['CI95_low_%']:.2f}, {r['CI95_high_%']:.2f}]"
        for _, r in bootstrap_df.iterrows()
    )
)
```

```python
# ============================================================
# CELL 39
# RELIABLE PER-SUBJECT ANALYSIS
# CELL 31 uses K=10 (0-7 genuine attempts per subject), which is too few.
# Here we use K=1 (window level) and require enough genuine windows.
# ============================================================

from scipy.stats import spearmanr

MIN_GENUINE_FOR_SUBJECT = 20

per = base["by_K"][1]["per_subject"]
enroll_counts = base["enroll_counts"]

rows = []

for s in range(N_SUBJECTS):

    if s not in per:
        continue

    g, i = per[s]["genuine"], per[s]["impostor"]

    if len(g) < MIN_GENUINE_FOR_SUBJECT or len(i) == 0:
        continue

    e, a = eer_auc(g, i)

    rows.append({
        "subject": subjects[s],
        "enroll_windows": int(enroll_counts[s]),
        "n_genuine": len(g),
        "EER_%": e * 100,
        "AUC": a,
    })

subject_reliable_df = pd.DataFrame(rows)

print("=" * 80)
print(f"PER-SUBJECT RESULTS, K=1, subjects with >= {MIN_GENUINE_FOR_SUBJECT} genuine windows")
print("=" * 80)

print("Subjects used     :", len(subject_reliable_df), "of", N_SUBJECTS)
print("Mean EER   (%)    :", round(subject_reliable_df["EER_%"].mean(), 2))
print("Median EER (%)    :", round(subject_reliable_df["EER_%"].median(), 2))
print("Std EER    (%)    :", round(subject_reliable_df["EER_%"].std(), 2))
print("Best / worst (%)  :",
      round(subject_reliable_df["EER_%"].min(), 2), "/",
      round(subject_reliable_df["EER_%"].max(), 2))

worse = subject_reliable_df[subject_reliable_df["AUC"] < 0.5]
print("\nSubjects with AUC < 0.5 (genuine scores BELOW impostor scores):")
display(worse.round(3))

rho, p = spearmanr(
    subject_reliable_df["enroll_windows"],
    subject_reliable_df["AUC"]
)
print(f"\nSpearman(enrollment windows, subject AUC) = {rho:.3f}  (p = {p:.4f})")

display(
    subject_reliable_df
    .sort_values("EER_%", ascending=False)
    .head(10)
    .round(3)
)

add_check(
    "Per-subject analysis uses enough genuine samples per subject",
    len(subject_reliable_df) > 0,
    f"{len(subject_reliable_df)} subjects with >= {MIN_GENUINE_FOR_SUBJECT} genuine windows",
    informational=True
)
add_check(
    "Subjects whose session-3 genuine scores fall below impostor scores",
    True,
    f"{len(worse)} subjects (session drift indicator)",
    informational=True
)
```

```python
# ============================================================
# CELL 40
# NEGATIVE CONTROL: LABEL PERMUTATION
# If we shuffle which subject each Session-3 window belongs to,
# the system must perform at chance (EER ~ 50%, AUC ~ 0.5).
# Anything else would mean a bug or leakage in the scoring code.
# ============================================================

N_PERM = 5

rng = np.random.default_rng(SEED)

perm_rows = []

for p in range(N_PERM):

    y_perm = rng.permutation(base["y_te"])

    g, i, _ = build_attempts(
        base["window_scores"],
        y_perm,
        K=1
    )

    e, a = eer_auc(g, i)

    perm_rows.append({"permutation": p, "EER_%": e * 100, "AUC": a})

perm_df = pd.DataFrame(perm_rows)

print("=" * 80)
print("LABEL-PERMUTATION CONTROL (K=1)")
print("=" * 80)
display(perm_df.round(4))

add_check(
    "Shuffled labels give chance-level performance (EER 45-55%, AUC 0.47-0.53)",
    bool(perm_df["EER_%"].between(45, 55).all()
         and perm_df["AUC"].between(0.47, 0.53).all()),
    f"mean EER = {perm_df['EER_%'].mean():.2f}%, mean AUC = {perm_df['AUC'].mean():.3f}"
)
```

```python
# ============================================================
# CELL 41
# OPEN-SET EVALUATION
# In CELL 26 the test impostors are the SAME people that appear as
# negatives while training every SVM. A real attacker is a stranger.
#
# Here, in every random split:
#   - 40 subjects are "known" (enrolled; SVMs train only on them)
#   - 20 subjects are "unseen" (never used in training at all)
#   closed-set impostors = other KNOWN subjects (seen as negatives)
#   open-set impostors   = UNSEEN subjects (never seen)
# Same genuine attempts, same models -> fair comparison.
# ============================================================

N_SPLITS = 5
N_KNOWN = 40

X_tr, y_tr, X_te, y_te = get_split(*FOLDS["S1+S2 -> S3"])

open_rows = []

for sp in range(N_SPLITS):

    r = np.random.RandomState(SEED + sp)
    order = r.permutation(N_SUBJECTS)

    known = sorted(order[:N_KNOWN].tolist())
    unseen = sorted(order[N_KNOWN:].tolist())

    tr_mask = np.isin(y_tr, known)

    targets = [s for s in known if (y_tr[tr_mask] == s).any()]

    models_k = train_models(
        X_tr[tr_mask],
        y_tr[tr_mask],
        targets,
        desc=f"Open-set split {sp}"
    )

    ws = score_windows(models_k, X_te)

    for K in K_LIST:

        g, imp_closed, _ = build_attempts(ws, y_te, K, impostor_ids=known)
        _, imp_open, _ = build_attempts(ws, y_te, K, impostor_ids=unseen)

        e_c, a_c = eer_auc(g, imp_closed)
        e_o, a_o = eer_auc(g, imp_open)

        open_rows.append({
            "split": sp,
            "K": K,
            "n_genuine": len(g),
            "n_imp_closed": len(imp_closed),
            "n_imp_open": len(imp_open),
            "EER_closed_%": e_c * 100,
            "EER_open_%": e_o * 100,
            "AUC_closed": a_c,
            "AUC_open": a_o,
        })

open_df = pd.DataFrame(open_rows)

open_summary = (
    open_df
    .groupby("K")
    .agg(
        EER_closed_mean=("EER_closed_%", "mean"),
        EER_closed_std=("EER_closed_%", "std"),
        EER_open_mean=("EER_open_%", "mean"),
        EER_open_std=("EER_open_%", "std"),
        AUC_closed=("AUC_closed", "mean"),
        AUC_open=("AUC_open", "mean"),
    )
    .round(3)
)

open_summary["gap_pp(open-closed)"] = (
    open_summary["EER_open_mean"] - open_summary["EER_closed_mean"]
).round(3)

print("=" * 80)
print(f"OPEN-SET vs CLOSED-SET  ({N_SPLITS} random splits, {N_KNOWN} known / "
      f"{N_SUBJECTS - N_KNOWN} unseen)")
print("=" * 80)
display(open_summary)

add_check(
    "Open-set EER vs closed-set EER (same models, same genuine attempts)",
    True,
    "gap per K (pp): " + ", ".join(
        f"K={k}: {v:+.2f}" for k, v in open_summary["gap_pp(open-closed)"].items()
    ),
    informational=True
)
```
